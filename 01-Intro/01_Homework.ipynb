```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/DataTalksClub/machine-learning-zoomcamp/main/cohorts/2026/data/car_fuel_efficiency_2026.csv")

fuel_types = df['fuel_type'].unique()

print(fuel_types)
```

```python
df
```

```python
# Count missing values in just the fuel_type column
missing_fuel_count = df['fuel_type'].isna().sum()
print(f"Missing fuel types: {missing_fuel_count}")

# Count missing values in EVERY column of the DataFrame
print(df.isna().sum())
```

```python
max_efficiency = df['fuel_efficiency_mpg'].max()
print(f"The maximum fuel efficiency is: {max_efficiency}")
```
